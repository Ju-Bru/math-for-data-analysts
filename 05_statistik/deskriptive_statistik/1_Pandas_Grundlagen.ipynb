## 6 – Zeilen auswählen mit `.iloc` und `.loc`

| Methode | Auswahl über | Merkhilfe |
|---|---|---|
| `.iloc[]` | die **Position** (0, 1, 2, ...) | *integer location* |
| `.loc[]` | das **Label** im Index und den Spaltennamen | *label* |

````python
# Die Zeile an Position 0, also die erste Zeile
umfrage.iloc[0]
````

````python
# Positionen 0 bis 2: Das Ende (3) ist ausgeschlossen, wie bei Listen
umfrage.iloc[0:3]
````

````python
# Ein einzelner Wert: Zeile an Position 0, Spalte an Position 2
print(umfrage.iloc[0, 2])
````

````python
# Labels 0 bis 3 und zwei Spalten nach Namen: Das Ende (3) ist hier eingeschlossen
umfrage.loc[0:3, ["kurszeit", "getraenk"]]
````

Auch `.loc` holt einen einzelnen Wert. Und `.iloc` nimmt Bereiche für Zeilen und Spalten gleichzeitig.

````python
# Ein einzelner Wert über Label und Spaltenname
print(umfrage.loc[0, "getraenk"])
````

````python
# Zeilen an Position 1 bis 3, Spalten an Position 0 und 1
umfrage.iloc[1:4, 0:2]
````


**Achtung:** Bei `.iloc` ist das Ende eines Bereichs ausgeschlossen, bei `.loc` ist es eingeschlossen.

Solange der Index einfach 0, 1, 2, ... lautet, bezeichnen Position und Label dieselbe Zeile. In Abschnitt 11 ändert sich das.

# Block C – Filtern und Spalten bearbeiten

## 7 – Zeilen filtern mit Bedingungen

Eine Bedingung auf einer Spalte ergibt eine Series aus `True` und `False`. Man nennt sie **Maske**.

````python
maske = umfrage["anfahrt_min"] > 45
print(maske.head(8))
````

Setzt du die Maske in eckige Klammern, bleiben nur die Zeilen mit `True` übrig.

````python
umfrage[maske]
````

Mehrere Bedingungen verbindest du mit `&` (und) oder `|` (oder). Jede Bedingung braucht eigene runde Klammern.

````python
# Vormittagskurs UND mindestens zwei Haustiere
umfrage[(umfrage["kurszeit"] == "Vormittag") & (umfrage["haustiere"] >= 2)]
````

`.isin([...])` prüft, ob ein Wert in einer Liste vorkommt. Das spart mehrere Bedingungen mit `|`.

````python
# Alle, die Tee oder Mate trinken
umfrage[umfrage["getraenk"].isin(["Tee", "Mate"])].head()
````

## 8 – Anteile berechnen

<img src="png%20Dateien/herr_bauchgefuehl.png" alt="Herr Bauchgefühl" width="150">

> **Herr Bauchgefühl:** „Die drei an der Kaffeemaschine lieben Kaffee. Also lieben alle Kaffee."

Er hat die Personen mit den IDs 101, 104 und 108 gefragt. Das ist seine Stichprobe.

````python
kaffeemaschine = umfrage[umfrage["teilnehmer_id"].isin([101, 104, 108])]
kaffeemaschine[["teilnehmer_id", "getraenk"]]
````

Dreimal Kaffee, in dieser Stichprobe also 100 %. Wer an der Kaffeemaschine fragt, trifft aber vor allem Kaffeetrinker. Das ist eine bequeme Auswahl und damit verzerrt.

Jetzt dieselbe Frage für alle 20 Befragten:

````python
trinkt_kaffee = umfrage["getraenk"] == "Kaffee"

print("Anzahl:", trinkt_kaffee.sum())
print("Anteil:", trinkt_kaffee.mean())
print(f"In Prozent: {trinkt_kaffee.mean():.0%}")
````

Warum funktioniert das? Pandas rechnet `True` als 1 und `False` als 0.

- `.sum()` zählt deshalb, wie oft die Bedingung zutrifft.
- `.mean()` liefert den Anteil, bei dem sie zutrifft.

Die Angabe `:.0%` im f-String formatiert die Zahl als Prozentwert ohne Nachkommastellen.

## 9 – Zufallsstichproben ziehen

`.sample(n=5)` zieht fünf zufällige Zeilen. Der Parameter `random_state` legt den Startwert des Zufallsgenerators fest, damit das Ergebnis wiederholbar ist.

````python
for startwert in [1, 2, 6]:
    stichprobe = umfrage.sample(n=5, random_state=startwert)
    anteil = (stichprobe["getraenk"] == "Kaffee").mean()
    print(f"Startwert {startwert}: {anteil:.0%} Kaffee")
````

Jede Stichprobe liefert einen anderen Anteil, obwohl unter allen 20 Befragten immer 40 % Kaffee trinken. Eine Zufallsstichprobe ist fair, aber fünf Personen sind für eine verlässliche Aussage zu wenig.

## 10 – Spalten hinzufügen, umbenennen und löschen

Eine neue Spalte entsteht durch eine Zuweisung. Pandas rechnet dabei Zeile für Zeile.

````python
# Anfahrt in Stunden statt in Minuten
umfrage["anfahrt_std"] = round(umfrage["anfahrt_min"] / 60, 2)
umfrage.head(3)
````

`.rename()` und `.drop()` geben einen **neuen** DataFrame zurück. Das Original ändert sich erst, wenn du das Ergebnis wieder zuweist.

````python
# rename erwartet ein Dictionary: {alter_name: neuer_name}
umbenannt = umfrage.rename(columns={"anfahrt_std": "anfahrt_stunden"})

print("umbenannt:", umbenannt.columns.tolist())
print("umfrage:  ", umfrage.columns.tolist())
````

````python
# Die Hilfsspalte wieder entfernen und das Ergebnis zuweisen
umfrage = umfrage.drop(columns=["anfahrt_std"])
print(umfrage.columns.tolist())
````

**Hinweis:** In fremdem Code siehst du oft `inplace=True`, zum Beispiel `umfrage.drop(columns=["anfahrt_std"], inplace=True)`. Das ändert den DataFrame direkt, ohne Zuweisung.

## 15 – Absolute und relative Häufigkeit

> **Herr Bauchgefühl:** „Acht Leute trinken Kaffee. Das ist die Mehrheit!"

`.value_counts()` zählt, wie oft jede Ausprägung vorkommt. Das ist die **absolute Häufigkeit**.

````python
print(umfrage["getraenk"].value_counts())
````

Mit `normalize=True` teilt Pandas jede Anzahl durch die Gesamtzahl. Das ist die **relative Häufigkeit**.

````python
print(umfrage["getraenk"].value_counts(normalize=True))
````

Kaffee ist die größte Gruppe, mit 40 % aber nicht die Mehrheit: 60 % trinken etwas anderes. Das zeigt erst die relative Häufigkeit. Herr Bauchgefühl stutzt.

`.value_counts()` sortiert nach Häufigkeit, die häufigste Ausprägung steht oben.

## 16 – Eine vollständige Häufigkeitstabelle

Für ordinale Merkmale kommt die **kumulierte Häufigkeit** dazu.

- `.sort_index()` sortiert nach der Ausprägung (S, M, L, XL) statt nach der Häufigkeit.
- `.cumsum()` zählt Zeile für Zeile auf.

````python
absolut = umfrage["tshirt"].value_counts().sort_index()

tabelle = pd.DataFrame({"absolut": absolut})
tabelle["relativ"] = tabelle["absolut"] / tabelle["absolut"].sum()
tabelle["kumuliert"] = tabelle["absolut"].cumsum()
tabelle["kumuliert_relativ"] = tabelle["relativ"].cumsum()
tabelle
````

Lesebeispiel: 12 von 20 Personen, also 60 %, tragen Größe M oder kleiner.

Kumulieren ist erst ab ordinalem Skalenniveau sinnvoll. Bei den Getränken gibt es keine Reihenfolge, auf die sich „bis einschließlich" beziehen könnte.

## 17 – Klassen bilden mit `pd.cut`

Ein stetiges Merkmal wie die Anfahrt hat fast so viele Ausprägungen wie Personen. `pd.cut` teilt die Werte in Klassen ein, `bins` nennt die Klassengrenzen.

Die Schreibweise kennst du vom Vormittag: `(20, 40]` bedeutet über 20 bis einschließlich 40.

````python
klassen = pd.cut(umfrage["anfahrt_min"], bins=[0, 20, 40, 60, 120])
print(klassen.value_counts().sort_index())
````

## 18 – Kreuztabellen mit `pd.crosstab`

> **Herr Bauchgefühl:** „Kaffee ist überall die Nummer eins, ob Vormittags- oder Abendkurs."

Eine Kreuztabelle zählt zwei Merkmale gleichzeitig aus. Das erste Merkmal bildet die Zeilen, das zweite die Spalten. `margins=True` ergänzt die Randhäufigkeiten.

````python
pd.crosstab(umfrage["kurszeit"], umfrage["getraenk"], margins=True, margins_name="Gesamt")
````

Im Abendkurs trinken 3 Personen Tee und nur 2 Kaffee. Weil die Kurse verschieden groß sind (8 und 12 Personen), vergleichst du besser **Zeilenprozente**: Jede Zeile ergibt zusammen 100 %.

````python
zeilenprozente = pd.crosstab(umfrage["kurszeit"], umfrage["getraenk"], normalize="index") * 100
zeilenprozente.round(1)
````

Im Abendkurs führt Tee mit 37,5 %, Kaffee kommt nur auf 25 %. Herr Bauchgefühl stutzt ein letztes Mal.

| Vom Vormittag | In Pandas |
|---|---|
| Zeilenprozente | `normalize="index"` |
| Spaltenprozente | `normalize="columns"` |
| Gesamtprozente | `normalize="all"` |

## 22 – Zusatz: Series und DataFrames selbst bauen

Bisher kam jede Series aus einer Spalte der Umfrage. Du kannst Series und DataFrames auch direkt anlegen. Die Beispiele stammen von der TechAcademy GmbH.

**Eine Series aus einer Liste.** Ohne weitere Angabe vergibt Pandas den Index 0, 1, 2, ...

````python
stundenlohn = pd.Series([55, 60, 48, 72, 65])
print(stundenlohn)
````

**Eine Series mit eigenem Index.** Mit `index=` bekommt jeder Wert ein eigenes Label. Über das Label holst du einen einzelnen Wert heraus.

````python
stundenlohn_id = pd.Series(
    [55, 60, 48, 72, 65],
    index=["T001", "T002", "T003", "T004", "T005"]
)

print(stundenlohn_id)
print()
print("Stundenlohn von T003:", stundenlohn_id["T003"])
````

**Ein DataFrame aus einem Dictionary.** Die Umfrage entstand aus einer Liste von Zeilen. Häufiger siehst du die Schreibweise mit einem Dictionary: Jeder Schlüssel ist ein Spaltenname, die Liste dahinter enthält die Werte dieser Spalte. So sind auch `bewerbungen` und `kurse` in Abschnitt 21 entstanden.

````python
trainer_daten = {
    "Name":            ["Sandra Meier", "Tobias Lange", "Yara Özcan", "Michael Brandt", "Nina Fuchs"],
    "Fachbereich":     ["Python & Data", "Datenbanken", "Linux & Netzwerk", "Python & Data", "Business Intelligence"],
    "Einstelldatum":   ["2019-03-01", "2021-06-15", "2020-09-01", "2022-01-10", "2023-04-01"],
    "Stundenlohn_EUR": [55, 60, 48, 72, 65],
}

trainer = pd.DataFrame(trainer_daten)
trainer
````

**Der Index gleich beim Erstellen.** Auch ein DataFrame nimmt `index=` an. Danach sprichst du Zeilen mit `.loc` über ihr Label an.

````python
trainer_ids = ["T001", "T002", "T003", "T004", "T005"]
trainer_mit_id = pd.DataFrame(trainer_daten, index=trainer_ids)

print(trainer_mit_id)
print()
print(trainer_mit_id.loc["T003"])
````

**Neue Spalten aus einer Liste und aus einer Bedingung.** Eine Liste braucht genau so viele Einträge, wie der DataFrame Zeilen hat. Eine Bedingung ergibt eine Spalte mit `True` und `False`.

````python
trainer["Kurse_pro_Monat"] = [8, 10, 6, 12, 9]
trainer["Senior"] = trainer["Stundenlohn_EUR"] >= 60
trainer
````

**Eine Spalte mit `del` löschen.** `del` entfernt die Spalte sofort und ohne Zuweisung. `drop(columns=...)` aus Abschnitt 10 ist flexibler, weil es mehrere Spalten auf einmal entfernen kann.

````python
del trainer["Kurse_pro_Monat"]
print(trainer.columns.tolist())
````

## 23 – Zusammenfassung

| Aufgabe | Befehl |
|---|---|
| Überblick | `head()`, `tail()`, `shape`, `columns`, `info()` |
| Spalte auswählen | `df["spalte"]`, `df[["a", "b"]]` |
| Kennzahlen | `min()`, `max()`, `sum()`, `count()`, `mean()` |
| Zeilen nach Position oder Label | `df.iloc[...]`, `df.loc[...]` |
| Filtern | `df[bedingung]`, `&`, `\|`, `isin([...])` |
| Anteil | `(bedingung).mean()` |
| Zufallsstichprobe | `sample(n=..., random_state=...)` |
| Spalten bearbeiten | Zuweisung, `rename(columns=...)`, `drop(columns=...)` |
| Index | `set_index()`, `reset_index()` |
| Datentypen und Ausprägungen | `dtypes`, `unique()`, `nunique()` |
| Geordnete Kategorie | `pd.Categorical(..., categories=..., ordered=True)` |
| Sortieren | `sort_values()`, `sort_index()` |
| Häufigkeiten | `value_counts()`, `value_counts(normalize=True)`, `cumsum()` |
| Klassen | `pd.cut(..., bins=...)` |
| Kreuztabelle | `pd.crosstab(a, b, margins=True, normalize=...)` |
| Series und DataFrame anlegen | `pd.Series([...], index=[...])`, `pd.DataFrame({...}, index=[...])` |
| Spalte sofort löschen | `del df["spalte"]` |
