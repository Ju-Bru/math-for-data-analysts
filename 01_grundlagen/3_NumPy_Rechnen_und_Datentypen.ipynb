# A – Brücke zur Mathe: Potenzen, Wurzeln, Logarithmen

## 1 – Die Mathe von heute Morgen in NumPy

| Mathe | NumPy | Beispiel |
|---|---|---|
| aⁿ | `a ** n` oder `np.power(a, n)` | `2 ** 3` → 8 |
| √a | `np.sqrt(a)` | `np.sqrt(16)` → 4 |
| ⁿ√a | `a ** (1/n)` | `27 ** (1/3)` → 3 |
| log₁₀(a) | `np.log10(a)` | `np.log10(1000)` → 3 |
| log₂(a) | `np.log2(a)` | `np.log2(8)` → 3 |
| ln(a) | `np.log(a)` | `np.log(np.e)` → 1 |
| eˣ | `np.exp(x)` | `np.exp(0)` → 1 |

**Achtung:** `np.log` ist der **natürliche** Logarithmus (Basis e), nicht log₁₀!

Das Besondere: Alle Funktionen arbeiten **mit ganzen Arrays**, nicht nur mit einzelnen Zahlen.

````python
zahlen = np.array([1, 4, 9, 16, 25])

print("Quadrat  :", zahlen ** 2)
print("Wurzel   :", np.sqrt(zahlen))
print("3. Wurzel:", np.array([8, 27, 64]) ** (1/3))
````

## 2 – Der Logarithmus als Umkehrung

Der Logarithmus beantwortet die Frage: **„Hoch wie viel?"**

- 2¹⁰ = 1024 → log₂(1024) = 10
- 10³ = 1000 → log₁₀(1000) = 3

Praktische Anwendung: log₁₀ verrät, **wie viele Stellen** eine Zahl hat.
Das ist nützlich, wenn Werte sehr unterschiedlich groß sind (3 Verkäufe vs. 12 000 Verkäufe).

````python
exponenten = np.arange(0, 11)
print(exponenten)
potenzen = 2 ** exponenten
print("2 hoch n :", potenzen)
print("log2     :", np.log2(potenzen))     # zurück zu den Exponenten

verkaeufe = np.array([3, 45, 820, 12000])
print("log10    :", np.log10(verkaeufe))
print("Stellen  :", np.floor(np.log10(verkaeufe)) + 1)      # nächstkleinere ganze Zahl
````

# B – Vektorisierung: Rechnen ohne Schleife

## 3 – Array und Zahl

Alle Rechenzeichen `+`, `-`, `*`, `/`, `**`, `//`, `%` wirken **elementweise**:
Die Zahl wird mit **jedem** Element verrechnet.

````python
preise = np.array([4.50, 3.20, 5.80, 2.90])   # Espresso, Crema, Filter, Kakao

print("+ 0,50 € :", preise + 0.50)
print("- 10 %   :", preise * 0.90)
print("in Cent  :", preise * 100)
````

## 4 – Zwei Arrays gleicher Form

Haben zwei Arrays die **gleiche Form**, wird **Element für Element** gerechnet: das erste mit dem ersten, das zweite mit dem zweiten usw.

````python
# Form = shape
mengen = np.array([120, 85, 60, 40])   # verkaufte Stück pro Produkt
print(preise)
print(np.shape(mengen))

umsatz = preise * mengen
print("Umsatz je Produkt:", umsatz)
````

Passen die Formen **nicht** zusammen, gibt es einen Fehler:

````python
try:
    preise * np.array([1, 2, 3])
except ValueError as e:
    print("Fehler:", e)
````

> 💡 **Kurz erwähnt:** Array und Zahl funktioniert, obwohl eine Zahl keine „Form" hat. Dahinter steckt **Broadcasting**, das lernst du in Notebook 4 genauer kennen.

## 5 – Warum ist das so schnell?

NumPy führt die Schleife **intern in C** aus, nicht in Python. Wir messen das mit `time.perf_counter()`:

````python
import time

liste = list(range(1_000_000))
arr = np.arange(1_000_000)

start = time.perf_counter()
ergebnis_liste = [x * 2 for x in liste]
zeit_liste = time.perf_counter() - start

start = time.perf_counter()
ergebnis_arr = arr * 2
zeit_arr = time.perf_counter() - start

print(f"Liste: {zeit_liste * 1000:.1f} ms")
print(f"Array: {zeit_arr * 1000:.1f} ms")
print(f"NumPy ist ca. {zeit_liste / zeit_arr:.0f}-mal schneller")
````

Die genauen Zahlen hängen von deinem Rechner ab, aber NumPy ist fast immer **deutlich** schneller.
Bei Millionen von Datensätzen macht das den Unterschied zwischen Sekunden und Minuten.

# C – Universelle Funktionen (ufuncs)

## 6 – Was ist eine ufunc?

Eine **ufunc** (*universal function*) ist eine NumPy-Funktion, die **elementweise** auf ein ganzes Array wirkt.
`np.sqrt`, `np.log10` oder `np.exp` aus Block A sind ufuncs.

| Gruppe | Beispiele |
|---|---|
| Mathematik | `np.sqrt`, `np.exp`, `np.log`, `np.log10`, `np.log2`, `np.power` |
| Runden | `np.round`, `np.floor`, `np.ceil`, `np.trunc` |
| Sonstiges | `np.abs`, `np.maximum`, `np.minimum` |

## 7 – Runden: vier Varianten

| Funktion | Bedeutung | 2.7 | -2.7 |
|---|---|---|---|
| `np.round(x, n)` | auf n Nachkommastellen runden auf nächste ganze Zahl | 3 | -3 |
| `np.floor(x)` | **ab**runden (nächstkleinere ganze Zahl) | 2 | -3 |
| `np.ceil(x)` | **auf**runden (nächstgrößere ganze Zahl) | 3 | -2 |
| `np.trunc(x)` | Nachkommastellen abschneiden | 2 | -2 |

Bei **negativen** Zahlen unterscheiden sich `floor` und `trunc`!

````python
werte = np.array([2.7, 2.2, -2.7, -2.2])

print("round:", np.round(werte))
print("floor:", np.floor(werte))
print("ceil :", np.ceil(werte))
print("trunc:", np.trunc(werte))

print("Preise auf 2 Stellen:", np.round(preise * 1.19, 2))
````

> 💡 **Kurz erwähnt:** `np.round` rundet bei genau x,5 zur **geraden** Zahl („Banker's Rounding"): `np.round(2.5)` → 2, `np.round(3.5)` → 4.
> Das ist in der Statistik üblich.

## 8 – Betrag, Maximum und Minimum

- `np.abs(x)`: Betrag, macht negative Werte positiv (z. B. Abweichungen)
- `np.maximum(a, b)`: vergleicht **Element für Element** und nimmt jeweils den größeren Wert
- `np.minimum(a, b)`: nimmt jeweils den kleineren

`np.maximum` ist praktisch für **Untergrenzen**: `np.maximum(werte, 0)` ersetzt alles unter 0 durch 0.

````python
soll = np.array([100, 80, 60, 50])
ist  = np.array([ 92, 85, 60, 41])

print("Abweichung     :", ist - soll)
print("Betrag         :", np.abs(ist - soll))
print("mind. 50 Stück :", np.maximum(ist, 50))
print("max. 90 Stück  :", np.minimum(ist, 90))
````

> 💡 **Kurz erwähnt – Operatoren sind auch ufuncs**
> `a + b` ist dasselbe wie `np.add(a, b)`, `a > b` dasselbe wie `np.greater(a, b)`.
> Die Funktionsschreibweise hat Zusatzoptionen, z. B. `out=`, um das Ergebnis direkt in ein vorhandenes Array zu schreiben und Speicher zu sparen.
> Im Alltag nimmst du aber die Operatoren.

> 💡 **Nicht verwechseln:** `np.maximum(a, b)` vergleicht zwei Arrays elementweise. `a.max()` sucht den größten Wert **in einem** Array, das kommt in Notebook 4.

# D – Datentypen (dtype)

## 9 – Was ist ein dtype?

Jedes Array hat **einen** Datentyp für **alle** Elemente. Er legt fest:

- welche Werte gespeichert werden können (ganze Zahlen, Kommazahlen, Wahrheitswerte, Text)
- welcher **Wertebereich** möglich ist
- wie viel **Speicher** jedes Element braucht

| Attribut | Bedeutung |
|---|---|
| `.dtype` | Datentyp |
| `.itemsize` | Bytes pro Element |
| `.nbytes` | Bytes für das ganze Array |

````python
ganz = np.array([1, 2, 3])
komma = np.array([1.5, 2.0, 3.25])
wahr = np.array([True, False, True])
text = np.array(["Espresso", "Crema"])

for a in [ganz, komma, wahr, text]:
    print(a.dtype, "| Bytes pro Element:", a.itemsize)
````

## 10 – Wertebereiche: hier kommt 2ⁿ ins Spiel

Ganze Zahlen werden in **Bits** gespeichert. Mit n Bits gibt es **2ⁿ verschiedene Werte**.

| dtype | Bits | Anzahl Werte | Wertebereich |
|---|---|---|---|
| `uint8` | 8 | 2⁸ = 256 | 0 bis 255 |
| `int8` | 8 | 2⁸ = 256 | -128 bis 127 |
| `uint16` | 16 | 2¹⁶ = 65 536 | 0 bis 65 535 |
| `int32` | 32 | 2³² ≈ 4,3 Mrd. | ca. ±2,1 Mrd. |
| `int64` | 64 | 2⁶⁴ | ca. ±9,2 Trillionen |

- `u` = *unsigned* = **ohne Vorzeichen**, nur Werte ≥ 0
- ohne `u`: die 256 Werte werden auf negativ und positiv aufgeteilt

Für Kommazahlen gibt es `float32` (ca. 7 gültige Stellen) und `float64` (ca. 15 Stellen, Standard).

Mit `np.iinfo` kannst du die Grenzen nachschlagen:

````python
for typ in [np.uint8, np.int8, np.uint16, np.int32]:
    info = np.iinfo(typ)
    print(f"{typ.__name__:7}: {info.min} bis {info.max}")

print("2 hoch 8 =", 2 ** 8, "| 2 hoch 16 =", 2 ** 16)
````

## 11 – dtype festlegen und umwandeln

**Beim Erstellen** mit `dtype=`:
```python
np.array([1, 2, 3], dtype=np.uint8)
np.zeros(5, dtype=np.int32)
```

**Nachträglich** mit `.astype()`. Wichtig: `astype` erzeugt immer eine **neue Kopie**, das Original bleibt unverändert.

````python
bestand = np.array([12, 40, 7], dtype=np.uint8)
print(bestand, bestand.dtype)

bestand_gross = bestand.astype(np.int32)
print(bestand_gross, bestand_gross.dtype)
print("Original:", bestand.dtype)          # unverändert
````

> 💡 **Kurz erwähnt – Speicher:** Ein `int64` braucht 8 Byte, ein `uint8` nur 1 Byte. Bei Millionen von Werten spart der kleinere Typ viel Arbeitsspeicher.
> Für unsere Übungsdaten spielt das keine Rolle.

> 💡 **Kurz erwähnt – Text:** In älteren Tutorials steht `np.string_` oder `np.unicode_`. Diese Namen gibt es seit **NumPy 2.0 nicht mehr**, heute heißen sie `np.bytes_` und `np.str_`.

# E – Typische Fallstricke

## 12 – Überlauf: wenn der Wertebereich nicht reicht

Wird beim Rechnen die Grenze eines Integer-Typs überschritten, **läuft der Wert über** und fängt von vorne an.
NumPy gibt dabei **keine Warnung** aus!

Stell dir einen Kilometerzähler vor, der nach 999 wieder bei 000 anfängt.

````python
lager = np.array([250, 100, 30], dtype=np.uint8)
print("Bestand      :", lager)
print("+ 10 geliefert:", lager + 10)     # 250 + 10 = 260 passt nicht in uint8!

print("- 50 verkauft:", lager - 50)     # 30 - 50 = -20 geht bei uint8 nicht!
````

Aus 260 wird 4 (260 − 256) und aus −20 wird 236 (−20 + 256). Die Rechnung ist falsch, und niemand merkt es.

**Lösung:** Rechtzeitig einen größeren Typ wählen, z. B. `lager.astype(np.int32) + 10`.

## 13 – Kommazahl → ganze Zahl: es wird abgeschnitten

`astype(int)` **rundet nicht**, sondern schneidet die Nachkommastellen ab (wie `np.trunc`).
Wer kaufmännisch runden will, muss **vorher** `np.round` anwenden.

````python
preise_cent = np.array([449.9, 319.6, 579.5])

print("astype    :", preise_cent.astype(int))
print("erst round:", np.round(preise_cent).astype(int))
````

## 14 – Automatische Typumwandlung (Upcasting)

Rechnet man mit verschiedenen Typen, wählt NumPy automatisch den **allgemeineren** Typ, damit nichts verloren geht:

| Kombination | Ergebnis |
|---|---|
| int + float | float |
| bool + int | int (True = 1, False = 0) |

Der zweite Fall ist sehr praktisch: Die **Summe einer Maske** zählt die `True`-Werte.

````python
ganz = np.array([1, 2, 3])
print((ganz + 0.5).dtype)            # int + float -> float

print(np.array([1, 2.5, 3]).dtype)   # gemischte Liste -> float

maske = mengen > 70
print(maske, "-> Anzahl True:", maske.sum())
````

## 15 – Text in Zahlen-Arrays und der Typ `object`

Steht in einer Liste auch nur **ein** Text, wird das ganze Array zu Text. Dann kann man nicht mehr rechnen.
Bei ganz gemischten Inhalten (z. B. mit `None`) fällt NumPy auf `object` zurück: Das Array speichert dann nur Verweise auf Python-Objekte, **alle Vorteile von NumPy sind weg**.

````python
gemischt = np.array([4.50, "3,20", 5.80])     # Komma statt Punkt -> Text!
print(gemischt, gemischt.dtype)

try:
    gemischt * 2
except Exception as e:
    print("Fehler:", type(e).__name__)

mit_luecke = np.array([4.50, None, 5.80])
print(mit_luecke, mit_luecke.dtype)
````

Wenn dein Array plötzlich `<U…` (Text) oder `object` als dtype hat, stimmt meist etwas mit den **Rohdaten** nicht.
Für fehlende Werte gibt es in NumPy `np.nan`, das lernst du in Notebook 4.

genau wie 1/3 im Dezimalsystem nicht (0,333…). Deshalb:

````python
print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)
print(np.isclose(0.1 + 0.2, 0.3))    # prüft "fast gleich"
````

**Regel:** Kommazahlen nie mit `==` vergleichen, sondern mit `np.isclose`.
