# A – Einstieg: Was ist NumPy?

## 1 – NumPy in einem Satz

**NumPy** steht für *Numerical Python*. Es ist eine Bibliothek, mit der man in Python schnell und bequem mit vielen Zahlen auf einmal rechnen kann.

Das Herzstück ist das **Array** (genauer: `ndarray` = *n-dimensionales Array*):
ein Gitter aus Werten, die **alle den gleichen Datentyp** haben, z. B. nur ganze Zahlen oder nur Kommazahlen.

**Warum nicht einfach Python-Listen?**

| | Python-Liste | NumPy-Array |
|---|---|---|
| Geschwindigkeit | langsam bei vielen Zahlen (Python-Schleifen) | sehr schnell (intern in C programmiert) |
| Speicher | jedes Element ist ein eigenes Python-Objekt | Zahlen liegen kompakt hintereinander im Speicher |
| Rechnen | nur mit Schleife | direkt mit dem ganzen Array (**Vektorisierung**) |
| Datentypen | beliebig gemischt | alle Elemente gleicher Typ |

**NumPy ist die Basis für fast alles in der Datenanalyse:** Pandas, Matplotlib, SciPy und scikit-learn bauen alle auf NumPy auf.
Wer NumPy versteht, versteht später auch Pandas viel leichter.

## 3 – Liste gegen Array: der Unterschied auf einen Blick

Kiez Kaffee hat drei Preise für Kaffeesorten (in €). Was passiert, wenn wir sie mit 2 multiplizieren?

````python
preise_liste = [4.50, 3.20, 5.80]
preise_array = np.array([4.50, 3.20, 5.80])

print("Liste * 2:", preise_liste * 2)   # Liste wird VERDOPPELT (aneinandergehängt)
print("Array * 2:", preise_array * 2)   # jede Zahl wird mit 2 multipliziert
````

Bei der Liste bräuchte man für „jeden Preis verdoppeln" eine Schleife:

````python
# Liste: Schleife nötig
doppelt = [p * 2 for p in preise_liste]
print(doppelt)

# Array: eine Zeile, keine Schleife
print(preise_array * 2)
````

Genau das ist die **Vektorisierung**: Eine Rechenoperation wird automatisch auf **jedes Element** angewendet.
Das ist kürzer, besser lesbar und bei großen Datenmengen viel schneller.

# B – Brücke zur Mathe: Boolesche Algebra

## 4 – Von der Wahrheitstabelle zum Bool-Array

Heute Vormittag hast du die Verknüpfungen **UND**, **ODER** und **NICHT** kennengelernt.
In NumPy gibt es dafür eigene Zeichen, die **elementweise** auf ganze Arrays wirken:

| Mathe | Python (einzelne Werte) | NumPy (Arrays) |
|---|---|---|
| A ∧ B (UND) | `a and b` | `a & b` |
| A ∨ B (ODER) | `a or b` | `a \| b` |
| ¬A (NICHT) | `not a` | `~a` |

Damit können wir die komplette Wahrheitstabelle in einem Schritt berechnen:

````python
A = np.array([True, True, False, False])
B = np.array([True, False, True, False])

print("A      :", A)
print("B      :", B)
print("A UND B:", A & B)    # T F F F
print("A ODER B:", A | B)   # T T T F
print("NICHT A:", ~A)       # F F T T
````

Jede Spalte der Wahrheitstabelle ist jetzt ein Array. NumPy rechnet alle vier Zeilen gleichzeitig aus.

Später benutzen wir genau diese Bool-Arrays, um Daten zu **filtern**, zum Beispiel „alle Tage mit Umsatz über 1000 € **UND** am Wochenende".

## 8 – Zufallszahlen

Zum Üben brauchen wir oft Beispieldaten. Die moderne Art in NumPy ist ein **Zufallsgenerator**:

```python
rng = np.random.default_rng(42)   # 42 = "Seed": gleiche Zahl -> gleiche Zufallszahlen
```

| Methode | Ergebnis |
|---|---|
| `rng.integers(low, high, size)` | ganze Zahlen von `low` bis **unter** `high` |
| `rng.random(size)` | Kommazahlen zwischen 0 und 1 |

Der **Seed** sorgt dafür, dass alle in der Klasse die gleichen „Zufalls"-Zahlen bekommen. So kann man Ergebnisse vergleichen.

````python
rng = np.random.default_rng(42)

print(rng.integers(1, 7, size=10))        # 10 Würfelwürfe
print()
print(rng.integers(0, 100, size=(2, 4)))  # 2x4-Array mit Zahlen von 0 bis 99
````

> 💡 **Kurz erwähnt:** In älteren Tutorials findest du oft `np.random.randint(...)` oder `np.random.rand(...)`.
> Das funktioniert noch, gilt aber als veraltet. Wir benutzen `default_rng`.

# D – Eigenschaften, Indexing und Slicing

## 9 – Wichtige Eigenschaften eines Arrays

Jedes Array „weiß" etwas über sich selbst:

| Attribut | Bedeutung | Beispiel für eine 2×3-Tabelle |
|---|---|---|
| `.ndim` | Anzahl der Dimensionen (Achsen) | `2` |
| `.shape` | Größe pro Dimension als Tupel `(Zeilen, Spalten)` | `(2, 3)` |
| `.size` | Gesamtzahl der Elemente | `6` |
| `.dtype` | Datentyp der Elemente | `int64` |

````python
# 2D: Umsätze von 2 Filialen (Zeilen) an 3 Tagen (Spalten)
umsatz_tabelle = np.array([
    [820, 950, 780],     # Kreuzberg
    [610, 700, 690],     # Neukölln
])

print("ndim :", umsatz_tabelle.ndim)
print("shape:", umsatz_tabelle.shape)
print("size :", umsatz_tabelle.size)
print("dtype:", umsatz_tabelle.dtype)
````

**Achtung bei der Shape:** Diese drei sehen ähnlich aus, sind aber verschieden:

| Shape | Bedeutung |
|---|---|
| `(3,)` | 1D-Array mit 3 Elementen (eine einfache Reihe) |
| `(1, 3)` | 2D-Array: **1 Zeile**, 3 Spalten |
| `(3, 1)` | 2D-Array: 3 Zeilen, **1 Spalte** |

````python
print(np.array([1, 2, 3]).shape)          # (3,)
print(np.array([[1, 2, 3]]).shape)        # (1, 3)
print(np.array([[1], [2], [3]]).shape)    # (3, 1)
````

## 10 – Indexing: einzelne Elemente

Wie bei Listen beginnt die Zählung bei **0**. Negative Indizes zählen von hinten.

Bei 2D-Arrays gibt man **Zeile, Spalte** an:

- `arr[0, 2]` → Zeile 0, Spalte 2 (empfohlene Schreibweise)
- `arr[0][2]` → funktioniert auch: erst Zeile 0 holen, dann daraus Element 2

````python
umsatz_woche = np.array([820, 950, 780, 1100, 1340])
print(umsatz_woche[0])     # erster Tag - 820
print(umsatz_woche[-1])    # letzter Tag - 1340

# 2D: Umsätze von 2 Filialen (Zeilen) an 3 Tagen (Spalten)
umsatz_tabelle = np.array([
    [820, 950, 780],     # Kreuzberg
    [610, 700, 690],     # Neukölln
])

print(umsatz_tabelle[1, 2])   # Neukölln, Tag 3 - 690
print(umsatz_tabelle[1][2])   # gleiches Ergebnis - 690
````

## 11 – Slicing: Teilbereiche ausschneiden

Schreibweise: `[start:stop:schritt]`, **stop ist nicht enthalten**.

Bei 2D-Arrays gibt es einen Slice für **jede Dimension**, getrennt durch ein Komma: `arr[zeilen, spalten]`.
Ein einzelner Doppelpunkt `:` bedeutet „alles".

| Ausdruck | Ergebnis |
|---|---|
| `arr[1:4]` | Elemente 1, 2, 3 |
| `arr[::2]` | jedes zweite Element |
| `arr2d[0, :]` | ganze Zeile 0 |
| `arr2d[:, 1]` | ganze Spalte 1 |
| `arr2d[:2, 1:]` | Zeilen 0–1, ab Spalte 1 |

````python
umsatz_woche = np.array([820, 950, 780, 1100, 1340])
print(umsatz_woche[1:4])      # Di bis Do - 950 780 1100
print(umsatz_woche[::2])      # Mo, Mi, Fr

# 2D: Umsätze von 2 Filialen (Zeilen) an 3 Tagen (Spalten)
umsatz_tabelle = np.array([
    [820, 950, 780],     # Kreuzberg
    [610, 700, 690],     # Neukölln
])
print(umsatz_tabelle[0, :])   # alle Tage von Kreuzberg
print(umsatz_tabelle[:, 1])   # Tag 2 von allen Filialen
````

> 💡 **Kurz erwähnt – Slices sind „Fenster", keine Kopien**
> Ein Slice zeigt auf **dieselben Daten** wie das Original (man nennt das eine *View*).
> Änderst du den Slice, änderst du auch das Original!
> Wenn du wirklich eine unabhängige Kopie brauchst: `.copy()` anhängen.

````python
original = np.array([10, 20, 30, 40])

fenster = original[0:2]       # View: 10 20
fenster[0] = 999              # 999 20  
print("Original nach Änderung am Slice:", original)   # 999 steht jetzt auch im Original!

original = np.array([10, 20, 30, 40])

kopie = original[0:2].copy()  # echte Kopie
print("Original nach Änderung an Kopie:", original)   # unverändert
````

## 12 – Vergleiche liefern Bool-Arrays

Vergleichsoperatoren (`>`, `<`, `>=`, `<=`, `==`, `!=`) wirken ebenfalls **elementweise**.
Das Ergebnis ist ein Array aus `True` und `False`, genau wie die Wahrheitswerte aus Block B.

## 13 – Boolean Indexing: mit einer Maske filtern

Ein Bool-Array nennt man in diesem Zusammenhang **Maske**. Setzt man die Maske in eckige Klammern,
bekommt man **nur die Elemente, bei denen `True` steht**:

```python
arr[maske]
```

````python
umsatz_woche = np.array([820, 950, 780, 1100, 1340])
maske = umsatz_woche > 900
print("Maske      :", maske)
print("Gefiltert  :", umsatz_woche[maske])

# Meist schreibt man es direkt in einer Zeile:
print("Kurzform   :", umsatz_woche[umsatz_woche > 900])
````

Bei 2D-Arrays funktioniert das genauso. Das Ergebnis ist dann immer ein **1D-Array** mit allen Treffern:

````python
# 2D: Umsätze von 2 Filialen (Zeilen) an 3 Tagen (Spalten)
umsatz_tabelle = np.array([
    [820, 950, 780],     # Kreuzberg
    [610, 700, 690],     # Neukölln
])
print(umsatz_tabelle[umsatz_tabelle < 800])
````

## 14 – Bedingungen kombinieren: `&`, `|`, `~`

Hier kommt die Boolesche Algebra zurück. Zwei Regeln sind wichtig:

1. Für Arrays **immer** `&`, `|`, `~` benutzen, **nicht** `and`, `or`, `not`.
2. Jede Einzelbedingung **in Klammern** setzen!

````python
umsatz_woche = np.array([820, 950, 780, 1100, 1340])

# Richtig: Klammern um jede Bedingung
print(umsatz_woche[(umsatz_woche > 800) & (umsatz_woche < 1200)])

# ODER: sehr niedrige oder sehr hohe Umsätze
print(umsatz_woche[(umsatz_woche < 800) | (umsatz_woche > 1200)])

# NICHT: alles außer über 1000
print(umsatz_woche[~(umsatz_woche > 1000)])
````

**Die typischen Fehler**, damit du die Meldung erkennst, wenn sie bei dir auftaucht:

````python
# Fehler 1: "and" statt "&"
try:
    umsatz_woche[(umsatz_woche > 800) and (umsatz_woche < 1200)]
except ValueError as e:
    print("Fehler 1:", e)

# Fehler 2: Klammern vergessen
try:
    umsatz_woche[umsatz_woche > 800 & umsatz_woche < 1200]
except ValueError as e:
    print("Fehler 2:", e)
````

Die Meldung *„The truth value of an array ... is ambiguous"* bedeutet fast immer: `and`/`or` benutzt oder Klammern vergessen.

## 15 – Werte per Maske ersetzen

Eine Maske kann auch links vom `=` stehen. Dann werden **nur die markierten Elemente** überschrieben:

````python
messwerte = np.array([12, -3, 8, -1, 15])
messwerte[messwerte < 0] = 0      # alle negativen Werte auf 0 setzen
print(messwerte)
````
