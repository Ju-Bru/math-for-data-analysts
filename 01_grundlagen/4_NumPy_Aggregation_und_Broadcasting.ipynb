**Heute:** Prozentrechnung mit Arrays · Summen und Durchschnitte (auch pro Zeile/Spalte) · laufende Summen und Wachstum · fehlende Werte (NaN) · Broadcasting

````python
import numpy as np

filialen = np.array(["Kreuzberg", "Neukölln", "Friedrichshain", "Wedding"])
tage = np.array(["Mo", "Di", "Mi", "Do", "Fr", "Sa", "So"])

# Umsätze einer Woche in €: Zeilen = Filialen, Spalten = Tage
woche = np.array([
    [820, 950, 780, 1100, 1340, 1500, 1200],
    [610, 700, 690,  720,  980, 1150,  900],
    [900, 870, 910,  950, 1210, 1380, 1250],
    [540, 600, 580,  610,  800,  950,  700],
])
print(np.shape(woche))
````

# A – Brücke zur Mathe: Prozent, Dreisatz, Wachstum

## 1 – Prozentrechnung mit Arrays

Die drei Grundformeln von heute Morgen gelten unverändert, nur rechnet NumPy sie für **alle Werte gleichzeitig**:

| Gesucht | Formel | NumPy |
|---|---|---|
| Prozentwert W | W = G · p / 100 | `G * p / 100` |
| Prozentsatz p | p = W / G · 100 | `W / G * 100` |
| Grundwert G | G = W · 100 / p | `W * 100 / p` |

````python
# Wochenziel je Filiale (Grundwert) und erreichter Umsatz am Montag (Prozentwert)
wochenziel = np.array([7000, 5500, 7000, 4500])
montag = woche[:, 0]

print(woche)
print("Montag in % vom Wochenziel:", np.round(montag / wochenziel * 100, 1))
print("20 % Rabatt auf die Montagsumsätze:", montag * 0.20)             # * 0,20
````

## 2 – Veränderung in Prozent und Wachstumsfaktor

- **Veränderung in %** = (neu − alt) / alt · 100
- **Wachstumsfaktor** q = 1 + p / 100 (bei +5 % ist q = 1,05, bei −5 % ist q = 0,95)
- **Wachstum über n Perioden:** Kₙ = K₀ · qⁿ

Mit einem Array von Exponenten bekommen wir die komplette Entwicklung in einer Zeile:

````python
montag_vorwoche = np.array([780, 640, 950, 500]) # "Kreuzberg", "Neukölln", "Friedrichshain", "Wedding"
print(montag)
print("Veränderung in %:", np.round((montag - montag_vorwoche) / montag_vorwoche * 100, 1))

# 5 % Wachstum pro Jahr, Startumsatz 250 000 €, Jahre 0 bis 5
jahre = np.arange(0, 6)
print("Umsatz:", np.round(250_000 * 1.05 ** jahre))
variable =  np.round(250_000 * 1.05 ** jahre)
print(np.array(variable * 1.05 ** jahre))
````

## 3 – Dreisatz

Der Dreisatz ist **Umrechnen über die Einheit**: Aus 250 g Bohnen macht man 12 Tassen, also aus 1 g 12/250 Tassen.
Mit einem Array kann man beliebig viele Mengen auf einmal umrechnen:

````python
gramm = np.array([250, 500, 1000, 75])
tassen = gramm * 12 / 250
print(tassen)
````

# B – Aggregationen: Daten zusammenfassen

## 4 – Ein ganzes Array zu einer Zahl machen

**Aggregieren** heißt: Viele Werte werden zu **einem** Wert zusammengefasst.

| Methode | Ergebnis |
|---|---|
| `arr.sum()` | Summe |
| `arr.mean()` | Mittelwert (Durchschnitt) |
| `arr.min()` / `arr.max()` | kleinster / größter Wert |
| `arr.argmin()` / `arr.argmax()` | **Position** des kleinsten / größten Werts |

Statt `arr.sum()` kann man auch `np.sum(arr)` schreiben, das Ergebnis ist gleich.

````python
kb = woche[0, :]  
print(kb.dtype)
# print(woche)    # Kreuzberg, ganze Woche
print(kb)
print(tage)


print("Summe       :", kb.sum())
print("Durchschnitt:", kb.mean())
print("Minimum     :", kb.min(), "am", tage[kb.argmin()])
print("Maximum     :", kb.max(), "am", tage[kb.argmax()])
````

`argmax` ist besonders nützlich: Mit der Position kann man im **passenden Namens-Array** nachschlagen, *wer* oder *wann* das Maximum hatte.

> 💡 **Kurz erwähnt:** Es gibt auch `arr.std()` (Standardabweichung) und `arr.var()` (Varianz). Was die bedeuten, lernst du in der Statistik-Woche.

## 5 – Der Parameter `axis`: pro Zeile oder pro Spalte?

Bei einer Tabelle will man meist **nicht** alles zu einer Zahl zusammenfassen, sondern **pro Filiale** oder **pro Tag**.
Dafür gibt es den Parameter `axis`:

```
                 Mo    Di    Mi    Do    Fr    Sa    So      axis=1  →  eine Summe pro ZEILE
Kreuzberg       820   950   780  1100  1340  1500  1200   →  7690
Neukölln        610   700   690   720   980  1150   900   →  5750
Friedrichshain  900   870   910   950  1210  1380  1250   →  7470
Wedding         540   600   580   610   800   950   700   →  4780
                 ↓     ↓     ↓     ↓     ↓     ↓     ↓
axis=0  →      2870  3120  2960  3380  4330  4980  4050      eine Summe pro SPALTE
```

**Merkhilfe:** `axis` gibt die Richtung an, in der zusammengefasst wird, und diese Achse **verschwindet**.

- `axis=0`: nach **unten** zusammenfassen → ein Wert **pro Spalte** (hier: pro Tag)
- `axis=1`: nach **rechts** zusammenfassen → ein Wert **pro Zeile** (hier: pro Filiale)

````python
print("Summe pro Filiale (axis=1):", woche.sum(axis=1))
print("Summe pro Tag     (axis=0):", woche.sum(axis=0))
print("Ø pro Tag             :", np.round(woche.mean(axis=0), 1))
print("stärkster Tag je Filiale  :", tage[woche.argmax(axis=1)])
````

## 6 – Masken zählen und Anteile berechnen

Aus Notebook 3 weißt du: `True` zählt als 1, `False` als 0. Damit wird es praktisch:

- `(bedingung).sum()` → **wie viele** Werte erfüllen die Bedingung?
- `(bedingung).mean()` → **welcher Anteil** erfüllt sie? (mal 100 = Prozent)

````python
ueber_1000 = woche > 1000
print(woche)

print("Anzahl Tage > 1000 € gesamt :", ueber_1000.sum())
print("Anzahl je Filiale           :", ueber_1000.sum(axis=1))
print(ueber_1000.mean())
print("Anteil in % gesamt          :", round(ueber_1000.mean() * 100, 1))
````

# C – Laufende Summen, Veränderungen, Wachstum

## 7 – `cumsum`: die laufende Summe

`np.cumsum` (*kumulierte Summe*) addiert Schritt für Schritt auf: Jede Stelle enthält die Summe **bis einschließlich** dieses Werts.

```
Umsatz :  820   950   780  1100 ...
cumsum :  820  1770  2550  3650 ...
```

Typische Frage: **Ab wann** ist ein Ziel erreicht?

````python
laufend = np.cumsum(kb)
print(kb)
print(laufend)

ziel = 5000
print("Ziel erreicht ab:", tage[laufend >= ziel][0])   # [0] = erster Treffer
````

## 8 – `np.diff`: Veränderung zum Vorwert

`np.diff(arr)` berechnet die Differenz zwischen **benachbarten** Werten. Das Ergebnis ist **ein Element kürzer**.
Die Veränderung in Prozent ist dann: Differenz / alter Wert · 100.

````python
print("Umsatz      :", kb)
print("Differenz   :", np.diff(kb))
print("Veränd. in %:", np.round(np.diff(kb) / * 100, 1))   # kb[:-1] = alle außer dem letzten
````

## 9 – `cumprod`: Wachstum mit wechselnden Raten

Wächst etwas jedes Jahr um einen **anderen** Prozentsatz, multipliziert man die Wachstumsfaktoren nacheinander:
K₃ = K₀ · q₁ · q₂ · q₃

Genau das macht `np.cumprod` (*kumuliertes Produkt*):

````python
raten = np.array([4, -2, 6, 3])        # Wachstum in % für 4 Jahre
faktoren = 1 + raten / 100
print("Faktoren  :", faktoren)
print("kumuliert :", np.round(np.cumprod(faktoren), 5))

start = 250_000
print("Umsatz    :", np.round(start * np.cumprod(faktoren)))
````

# D – Fehlende Werte und Ausreißer

## 10 – `np.nan`: ein Platzhalter für „fehlt"

Erinnerst du dich an Notebook 1? Dort haben wir Buchungsfehler auf **0** gesetzt. Aber 0 € Umsatz ist etwas anderes als „keine Daten"!
Für fehlende Werte gibt es `np.nan` (*Not a Number*).

Drei Dinge musst du wissen:

1. `np.nan` ist eine **Kommazahl**. Ein Integer-Array kann kein NaN enthalten, man braucht `float`.
2. NaN ist **ansteckend**: Jede Rechnung mit NaN ergibt NaN.
3. NaN ist **nicht einmal gleich sich selbst**: `np.nan == np.nan` ist `False`. Zum Prüfen gibt es `np.isnan()`.

````python
kasse = np.array([820, np.nan, 780, 1100, np.nan, 1500, 1200])
print(kasse, kasse.dtype)

print("Summe          :", kasse.sum())         # NaN steckt an
print("nan == nan     :", np.nan == np.nan)
print("isnan          :", np.isnan(kasse))
print("Anzahl fehlend :", np.isnan(kasse).sum())
print(len(kasse))
````

## 11 – Mit NaN rechnen

| Ziel | Lösung |
|---|---|
| NaN beim Rechnen ignorieren | `np.nansum`, `np.nanmean`, `np.nanmin`, `np.nanmax` … |
| NaN herausfiltern | `arr[~np.isnan(arr)]` |
| NaN durch einen Wert ersetzen | `np.where(np.isnan(arr), ersatz, arr)` |

````python
kasse = np.array([820, np.nan, 780, 1100, np.nan, 1500, 1200])
print("nansum   :", np.nansum(kasse))
print("nanmean  :", round(np.nanmean(kasse), 1))
print("gefiltert:", kasse[~np.isnan(kasse)])

ersetzt = np.where(np.isnan(kasse), np.nanmean(kasse), kasse)
print("ersetzt  :", np.round(ersetzt))
````

**Achtung:** Ob man NaN ignoriert, herausfiltert oder ersetzt, ist eine **fachliche** Entscheidung, keine technische.
Wer fehlende Tage mit dem Durchschnitt füllt, tut so, als hätte es diese Umsätze gegeben.

## 12 – `np.clip`: Werte auf einen Bereich begrenzen

Der Röstofen misst die Temperatur. Manchmal liefert der Sensor unsinnige Werte (z. B. 0 °C oder 999 °C).
Bekannt ist: Realistisch sind nur Werte **zwischen 180 und 240 °C**.

`np.clip(arr, untergrenze, obergrenze)`:
- alles **unter** der Untergrenze wird zur Untergrenze
- alles **über** der Obergrenze wird zur Obergrenze
- alles dazwischen bleibt unverändert

````python
temperatur = np.array([205, 212, 0, 219, 226, 999, 231, 228, 175])

print("roh    :", temperatur)
print("clip   :", np.clip(temperatur, 180, 240))
````

`np.clip` macht dasselbe wie zwei verschachtelte `np.where`, ist aber kürzer und lesbarer:

````python
per_where = np.where(temperatur < 180, 180, np.where(temperatur > 240, 240, temperatur))
print(per_where)
````

> 💡 **Kurz erwähnt:** Ob man Ausreißer **begrenzt** (clip) oder **als fehlend markiert** (NaN), hängt wieder von der Fachfrage ab.
> Ein Sensor, der 999 °C meldet, hat eigentlich *gar nichts* gemessen. Dann wäre NaN ehrlicher.

# E – Broadcasting: Arrays unterschiedlicher Form verrechnen

## 13 – Warum funktioniert `arr + 4`?

Eigentlich rechnet NumPy nur Arrays **gleicher Form** Element für Element.
Trotzdem funktioniert `np.array([1, 2, 3, 4]) + 4`. Der Grund: NumPy **streckt** die 4 gedanklich auf die passende Form:

```
[1, 2, 3, 4]        [1, 2, 3, 4]
     +         →         +
     4              [4, 4, 4, 4]
```

Dieses automatische Strecken heißt **Broadcasting**. Es funktioniert nicht nur mit einzelnen Zahlen, sondern auch mit ganzen Zeilen und Spalten.

## 14 – Eine Zeile auf eine Tabelle anwenden

`woche` hat die Form `(4, 7)`. Ein Array mit **7** Werten (eines pro Tag) wird auf **jede Zeile** angewendet:

```
woche (4, 7)                     zuschlag (7,)
[ 820  950 ... 1200]      +      [0 0 0 0 0 50 50]    ← wird für jede Zeile wiederholt
[ 610  700 ...  900]      +      [0 0 0 0 0 50 50]
[ 900  870 ... 1250]      +      [0 0 0 0 0 50 50]
[ 540  600 ...  700]      +      [0 0 0 0 0 50 50]
```

````python
zuschlag = np.array([0, 0, 0, 0, 0, 50, 50])   # Wochenend-Aktion: +50 € pro Filiale
print(woche + zuschlag)
print()

# Abweichung jeder Filiale vom Tagesdurchschnitt aller Filialen
tages_schnitt = woche.mean(axis=0)            # Form (7,)
print(tages_schnitt)
print()
print(np.round(woche - tages_schnitt))
````

## 15 – Eine Spalte auf eine Tabelle anwenden

Jetzt wollen wir **pro Filiale** rechnen, z. B. den Anteil jedes Tages am **Wochenumsatz der Filiale**.
Die Wochensumme hat 4 Werte, einen pro Zeile. Aber:

````python
summe_filiale = woche.sum(axis=1)
print(summe_filiale.shape)

try:
    woche / summe_filiale
except ValueError as e:
    print("Fehler:", e)
````

NumPy vergleicht die Formen **von rechts**: `(4, 7)` und `(4,)` → 7 und 4 passen nicht.
Wir müssen aus den 4 Werten eine **Spalte** machen, also die Form `(4, 1)`. Dafür gibt es zwei Wege:

- `keepdims=True` beim Aggregieren: die zusammengefasste Achse bleibt mit Länge 1 erhalten
- `[:, np.newaxis]`: fügt nachträglich eine Achse der Länge 1 hinzu

```
woche (4, 7)                         summe (4, 1)
[ 820  950 ... 1200]       /         [7690]    ← wird nach rechts wiederholt
[ 610  700 ...  900]       /         [5750]
[ 900  870 ... 1250]       /         [7470]
[ 540  600 ...  700]       /         [4780]
```

````python
summe_spalte = woche.sum(axis=1, keepdims=True)
print(summe_spalte.shape)

anteil = woche / summe_spalte * 100
print(np.round(anteil, 1))
print("Kontrolle, jede Zeile = 100 %:", np.round(anteil.sum(axis=1)))

# gleiches Ergebnis mit np.newaxis
print(summe_filiale)
print(summe_filiale[:, np.newaxis].shape)
````

**Kurzregel für Broadcasting:** Vergleiche die Formen **von rechts nach links**. Zwei Längen passen, wenn sie **gleich** sind oder eine davon **1** ist.

| Form A | Form B | passt? |
|---|---|---|
| `(4, 7)` | `(7,)` | 7 = 7 |
| `(4, 7)` | `(4, 1)` | 7 und 1, 4 = 4 |
| `(4, 7)` | `(4,)` | 7 ≠ 4 |

> 💡 **Kurz erwähnt:** Die offiziellen Broadcasting-Regeln in der NumPy-Dokumentation sind etwas formaler (fehlende Achsen werden links mit 1 aufgefüllt usw.). Für unsere Zwecke reicht die Kurzregel.
